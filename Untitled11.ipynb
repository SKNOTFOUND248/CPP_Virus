```python
class Meta(type):

    def __new__(cls, name, bases, dct):
        print("Creating class:", name,"i am in __new__")
        print(dct)
        return super().__new__(cls, name, bases, dct)
    
    def __init__(cls, name, bases, dct):
        print("hello i in __init__")
        return super().__init__(cls)
```

```python
a = Meta.__call__(Meta,"A",(B,),{'x':10,"helo":lambda self,co : print(co)})

s = a()
```

```python
# finsd the second larges number in list 

n = [1,2]

max =n[0]
second = -1
third = -1

for i in n:
    if i > max:
        second = max
        max = i
    elif i > second and i != max:
        third = second
        second = i
    elif i > third and i != second and i != max :
        third = i

print(third)
        
```

```python
def fibonacc(n):
    if n==0 or n==1:
        return n
    
    return fibonacc(n-1) + fibonacc(n-2)

print(fibonacc(10))
```

```python
def valid(n):
    if n == 2 or n == 3:
        return True
    if n==0 or n==1 or n%2==0 or n%3==0:
        return False
    
    for i in range(5,n,2):
        if n%i==0:
            return False   
        
    return True

n=int(input("enter a number"))    
m = int(input("enter  a number "))

for i in range(n,m):
    if valid(i):
        print(i)
```
